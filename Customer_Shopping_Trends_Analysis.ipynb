```python
import pandas as pd

# 1. Sourcing and loading the retail dataset
df = pd.read_csv('shopping_trends_updated.csv')

# 2. Inspecting the dataset shape (Rows vs Columns)
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")

# 3. Data Integrity Check: Assessing data fields for missing/null values
print("\nMissing values check:")
print(df.isnull().sum())
```

```python
# 4. Aggregating purchase amount by category to find top revenue drivers
category_sales = df.groupby('Category')['Purchase Amount (USD)'].sum().sort_values(ascending=False)

print("Top Selling Categories (by Revenue):")
print(category_sales)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plotting Category Sales for exploratory data analysis (EDA)
plt.figure(figsize=(10,6))

# Assigning x to hue and setting legend=False to fix the FutureWarning
sns.barplot(x=category_sales.index, y=category_sales.values, hue=category_sales.index, palette='viridis', legend=False)

# Adding labels and styling
plt.title('Total Sales by Category', fontsize=14, fontweight='bold')
plt.ylabel('Total Revenue (USD)', fontsize=12)
plt.xlabel('Category', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Rendering the plot
plt.show()
```
